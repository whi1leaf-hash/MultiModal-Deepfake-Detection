```python
import torch
import os
from transformers import BertTokenizer, BertModel
import numpy as np
```

```python
def sentence2vector(sentence, model, tokenizer):
    # 将句子编码为BERT输入所需的格式
    inputs = tokenizer(sentence, return_tensors="pt")
    
    # 获取BERT模型的输出
    with torch.no_grad():
        outputs = model(**inputs)
    
    # 提取"[CLS]"标记的输出作为句向量
    sentence_vector = outputs.last_hidden_state[:, 0, :].squeeze().numpy()

    return sentence_vector
```

```python
# 数据集地址
DATASET_DIR = os.environ.get("FAKEAVCELEB_DIR", "data/FakeAVCeleb")
listType = os.listdir(DATASET_DIR)  # 列出文件夹下所有的目录与文件

# 预训练模型地址
BERT_MODEL_DIR = "bert-base-uncased"
tokenizer = BertTokenizer.from_pretrained(BERT_MODEL_DIR)
model = BertModel.from_pretrained(BERT_MODEL_DIR)
```

```python
# 示例用法
# 例句
sentence = "This is an example sentence."

# 获取句向量
sentence_vector = sentence2vector(sentence, model, tokenizer)

# 现在sentence_vector就是句子的向量表示
print(sentence_vector)
print(sentence_vector.shape)
print(type(sentence_vector))
```
