## Import all the essential libraries

```python
import torch
import librosa
import numpy as np
import numpy.typing
import soundfile as sf
from scipy.io import wavfile
from IPython.display import Audio
from transformers import Wav2Vec2ForCTC, Wav2Vec2Tokenizer, Wav2Vec2Processor
```

## Load tokenizer and model

```python
WAV2VEC_MODEL_DIR = "wav2vec2-base-960"
tokenizer = Wav2Vec2Processor.from_pretrained(WAV2VEC_MODEL_DIR)
```

```python
model = Wav2Vec2ForCTC.from_pretrained(WAV2VEC_MODEL_DIR)
```

## Select a sample audio

```python
file_name1 = 'sa1_1.wav'
file_name2 = 'sa1_2.wav'
```

## Adjust sample rate and output

```python
data = wavfile.read(file_name1)
framerate = data[0]
sounddata = data[1]
time = np.arange(0,len(sounddata))/framerate
input_audio, _ = librosa.load(file_name1, sr=16000)
input_values1 = tokenizer(input_audio, return_tensors="pt").input_values
input_values1
```

```python
data = wavfile.read(file_name2)
framerate = data[0]
sounddata = data[1]
time = np.arange(0,len(sounddata))/framerate
input_audio, _ = librosa.load(file_name2, sr=16000)
input_values2 = tokenizer(input_audio, return_tensors="pt").input_values
input_values2
```

```python
logits1 = model(input_values1).logits
logits1
```

```python
logits2 = model(input_values2).logits
logits2
```

```python
predicted_ids1 = torch.argmax(logits1, dim=-1)
predicted_ids1
```

```python
transcription1 = tokenizer.batch_decode(predicted_ids1)[0]
transcription1
```

```python
predicted_ids2 = torch.argmax(logits2, dim=-1)
transcription2 = tokenizer.batch_decode(predicted_ids2)[0]
transcription2
```
